```python
class railone:
    irctc =  "Indian Railway"                  # class variable/attribute
    def __init__(self,train_name,passenger_name,gender,age):      # define constructor /special method/ special function/instance method
        self.train_name = train_name                               # instance variable/attribute1
        self.passenger_name = passenger_name                        # instance variable/attribute2
        self.gender = gender                                       # instance variable/attribute3
        self.age = age                                          # instance variable/attribute4

 # define class methods
    def passenger_info(self):
        self.address = input("Enter the address: ")
        print(f'passenger_name is {self.passenger_name}')
        print(f'age is {self.age}')
        print(f'gender is {self.gender}')
        print(f'address is {self.address}')

     #define class methods
    def booking_info(self):
        self.booking_available = 10
        self.fare = 1000
        self.booking = int(input("Enter the booking:"))
        self.choice = input("Enter the choice:").lower()
        if self.choice == "firstclass":

         if (self.booking <= self.booking_available):
            self.fare = self.fare * self.booking
            self.booking_available = self.booking_available - self.booking
            print(f'booking_info is {self.booking}')
            print(f'train name is {self.train_name}')
            print(f'passenger_name is {self.passenger_name}')
            print(f'fare is {self.fare}')
            print(f'booking_available is {self.booking_available}')
        else:
            print('booking is not available')

train_name = input('Enter the name of the train: ')
passenger_name = input('Enter the name of the passenger:')
age = int(input('Enter the age:'))
gender = input('Enter the gender:')

railone1 = railone(train_name,passenger_name,gender,age)

railone1.passenger_info()
railone1.booking_info()
```
