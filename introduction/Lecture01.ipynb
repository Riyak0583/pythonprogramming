```python
# Write a program wheather the entered string is palindrome or not


string = input ('Enter the string :')

string1 = string[ : : -1]  # reverse the given string

string==string1

if string == string1:
    print ("The given string is a Palindrom ")
else :
    print("The given string is not a Palindrom")
```
